# Initialize Gemini API

```python
from google import genai
client = genai.Client()

def read_sas():
    try:
        with open("../data/Project_1/SAS Programs/0 - Data and Measure Standardization_2025Jul.sas", "r", encoding='utf8') as file:
            content = file.read()
            return content
    except FileNotFoundError:
        return f"Error: The file path was not found."

output = read_sas()

translator = client.interactions.create(
    model="gemini-3.8-flash",
    input="Translate the SAS code into Python using the provided file." + output,
)
print(translator.output_text)
```
